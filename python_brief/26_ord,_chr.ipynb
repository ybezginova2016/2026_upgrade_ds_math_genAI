### Узнать, что это за числа, по которым происходит сравнение, можно с помощью функции ord(), которая принимает символ и возвращает его номер в кодировочной таблице.

```python
print(')' > 'A')  # False
```

```python
print(ord(')'), ord('A'))  # 41 65
```

```python
print(chr(65))
print(ord('A'))
```

```python
# Напишите программу, которая найдёт сумму кодов всех пяти введённых символов.

s1 = input()
s2 = input()
s3 = input()
s4 = input()
s5 = input()

print(ord(s1) + ord(s2) + ord(s3) + ord(s4) + ord(s5))
```
