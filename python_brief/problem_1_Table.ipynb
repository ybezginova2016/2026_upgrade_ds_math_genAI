## Problem Statement

Conduct a small study of the input strings and write the results as a table.

First, a substring is entered for checking, and then three more strings are entered.

First, print the table header:
`String<spaces for alignment up to 10 characters>:number`

Then print a table where, for each of the three input strings, after the colon you should write:
- `1` if the first entered substring is contained in this string;
- `10` if the substring is not contained in this string;
- `111` if the whole string exactly matches the entered substring.

Use `10` characters for the left part of the table, not counting the colon. Use `5` characters for the number.


```python
pattern = input()
s1 = input()
s2 = input()
s3 = input()

print(f'{"Строка":10}:число')

if s1 == pattern:
    value_1 = 111
elif pattern in s1:
    value_1 = 1
else:
    value_1 = 10

if s2 == pattern:
    value_2 = 111
elif pattern in s2:
    value_2 = 1
else:
    value_2 = 10

if s3 == pattern:
    value_3 = 111
elif pattern in s3:
    value_3 = 1
else:
    value_3 = 10

print(f'{s1:10}:{value_1:5}')
print(f'{s2:10}:{value_2:5}')
print(f'{s3:10}:{value_3:5}')
```
