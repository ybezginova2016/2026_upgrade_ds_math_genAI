```python
"""
Проверьте, не встречаются ли в одной строке мальчик и лев? Если есть только мальчик Лев, то выведите:
Мальчик есть, льва нет.

Если только лев, то:
Есть лев, мальчика нет.

Если никого нет, то:
Никого.

Если есть оба, то:
Лев, беги!

Пример
Ввод	Вывод
Из рассказа Лев узнал, что лев - это хищное животное.
Лев, беги!
"""
```

```python
s = str(input())

if 'Лев' in s and 'лев' not in s:
    print('Мальчик есть, льва нет.')
elif 'лев' in s and 'Лев' not in s:
    print('Есть лев, мальчика нет.')
elif 'лев' not in s and 'Лев' not in s:
	print('Никого.')
else:
    print('Лев, беги!')
```
