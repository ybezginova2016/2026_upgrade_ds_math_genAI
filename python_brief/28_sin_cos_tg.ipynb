```python
"""
В прямоугольном треугольнике даны один из углов (непрямой) и гипотенуза (h). Напишите программу, которая выведет катет (a), прилежащий к углу.

Угол вводится в градусах.
"""

from math import cos, radians

# функции sin, cos, tan в Python из модуля math принимают угол в радианах, а не в градусах

alpha = float(input())
h = float(input())

a = h * cos(radians(alpha))
print(a)
```
