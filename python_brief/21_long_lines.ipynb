```python
n = int(input())

print(
    (n // 10000) % 2 == 0 and
   (n // 1000 % 10) % 2 == 0 and
   (n // 100 % 10) % 2 == 0 and
   (n // 10 % 10) % 2 == 0 and
   (n % 10) % 2 == 0
)
```

```python
"""
Для трёх введённых чисел определите, есть ли среди них такая пара, что остаток от деления на 7 одного числа равен целой части от деления на 2 другого числа? Если такая пара есть, выведите True, если нет, то False.
"""
# l = 43
# n = 746
# m = 8879

l = int(input())
n = int(input())
m = int(input())

print(
    # l, n
    l % 7 == n // 2 or
    # n, m
    n % 7 == m // 2 or
    # l, m
    l % 7 == m // 2
)

```
