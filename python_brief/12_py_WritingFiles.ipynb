## Write and Save Files in Python

```python
# Write line to file
exmp2 = '/Example2.txt'
with open(exmp2, 'w') as writefile:
    writefile.write("This is line A")
```

```python
# Read file

with open(exmp2, 'r') as testwritefile:
    print(testwritefile.read())
```
