```python
"""
Произведение крайних

Среди пяти введённых чисел найдите самое маленькое и самое большое и выведите их произведение. Использовать условный оператор запрещено!
"""

a = int(input())
b = int(input())
c = int(input())
d = int(input())
e = int(input())

print(max(a, b, c, d, e) * min(a, b, c, d, e))
```

```python
"""
Произведение крайних (по модулю)

Среди пяти введённых чисел найдите самое маленькое и самое большое по модулю и выведите их модуль их произведения. Использовать условный оператор запрещено!
"""

a = int(input())
b = int(input())
c = int(input())
d = int(input())
e = int(input())

print(max(abs(a), abs(b), abs(c), abs(d), abs(e)) * min(abs(a), abs(b), abs(c), abs(d), abs(e)))
```

```python
a = int(input())
b = int(input())
c = int(input())
d = int(input())
e = int(input())

min_abs = min(abs(a), abs(b), abs(c), abs(d), abs(e))
max_abs = max(abs(a), abs(b), abs(c), abs(d), abs(e))

print(abs(min_abs * max_abs))
```

```python
"""
В эксперименте иногда бывают случайные отклонения: оборудование виновато, датчики не так сработали, под окнами лаборатории проехал тяжёлый грузовик…

Напишите программу, которая отбросит два крайних значения — наибольшее и наименьшее, а для остальных найдёт среднее.

Формат ввода
Вводится пять чисел — целых или вещественных.

Формат вывода
Выведите среднее трёх из них, отбросив наименьшее и наибольшее из введённых пяти. Ответ округлите до двух знаков после запятой.
"""

a = float(input())
b = float(input())
c = float(input())
d = float(input())
e = float(input())

max_n = max(a, b, c, d, e)
min_n = min(a, b, c, d, e)

print((a + b + c + d + e - min_n - max_n) / 3)
```

```python
a = float(input())
b = float(input())
c = float(input())
d = float(input())
e = float(input())

mean = (a + b + c + d + e - min(a, b, c, d, e) - max(a, b, c, d, e)) / 3
print(round(mean, 2))
```
