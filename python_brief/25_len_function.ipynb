```python
"""
Среди пяти введённых строк с разной длиной выберите самую длинную и самую короткую.
Выведите: сначала самую короткую, а затем сколько раз длина самой
короткой помещается в самой длинной.

"""
s1 = input()
s2 = input()
s3 = input()
s4 = input()
s5 = input()

l_s1 = len(s1)
l_s2 = len(s2)
l_s3 = len(s3)
l_s4 = len(s4)
l_s5 = len(s5)

min_l = min(l_s1, l_s2, l_s3, l_s4, l_s5)
max_l = max(l_s1, l_s2, l_s3, l_s4, l_s5)

if l_s1 == min_l:
    print(s1)
elif l_s2 == min_l:
    print(s2)
elif l_s3 == min_l:
    print(s3)
elif l_s4 == min_l:
    print(s4)
else:
    print(s5)

print(max_l // min_l)
```
