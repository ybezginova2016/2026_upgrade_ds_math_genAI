```python
n = 999
```

```python
for i in range(9, 1, -1):
    if n % i == 0 and n != i:
        print(i)
        break
else:
    print(-1)
```

```python
if n % 9 == 0 and n != 9:
    print(9)
elif n % 8 == 0 and n != 8:
    print(8)
elif n % 7 == 0 and n != 7:
    print(7)
elif n % 6 == 0 and n != 6:
    print(6)
elif n % 5 == 0 and n != 5:
    print(5)
elif n % 4 == 0 and n != 4:
    print(4)
elif n % 3 == 0 and n != 3:
    print(3)
elif n % 2 == 0 and n != 2:
    print(2)
else:
    print(-1)
```
