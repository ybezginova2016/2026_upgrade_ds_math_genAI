```python
"""
Напишите программу, которая проверит, синхронизированы ли между собой орбиты двух планет, что означает: есть ли такое число от 2 до 9, на которое можно разделить обе орбиты?

Если есть, то для наибольшего такого числа ваша программа должна вывести фразу:
Орбиты относятся как <число1>:<число2>.

В этой строке <число1> означает целую часть от деления большего из чисел на найденный делитель, а <число2> — целую часть от деления на найденный делитель оставшегося числа.

Если такого делителя найти не удалось, то выведите:
Орбиты не синхронизированы.

Формат ввода
Вводятся два числа, большие 9, первое больше второго.
"""

orbit_1 = 90
orbit_2 = 56
```

```python
if orbit_1 % 9 == 0 and orbit_2 % 9 == 0:
    print(f'Орбиты относятся как {orbit_1 // 9}:{orbit_2 // 9}.')
elif orbit_1 % 8 == 0 and orbit_2 % 8 == 0:
    print(f'Орбиты относятся как {orbit_1 // 8}:{orbit_2 // 8}.')
elif orbit_1 % 7 == 0 and orbit_2 % 7 == 0:
    print(f'Орбиты относятся как {orbit_1 // 7}:{orbit_2 // 7}.')
elif orbit_1 % 6 == 0 and orbit_2 % 6 == 0:
    print(f'Орбиты относятся как {orbit_1 // 6}:{orbit_2 // 6}.')
elif orbit_1 % 5 == 0 and orbit_2 % 5 == 0:
    print(f'Орбиты относятся как {orbit_1 // 5}:{orbit_2 // 5}.')
elif orbit_1 % 4 == 0 and orbit_2 % 4 == 0:
    print(f'Орбиты относятся как {orbit_1 // 4}:{orbit_2 // 4}.')
elif orbit_1 % 3 == 0 and orbit_2 % 3 == 0:
    print(f'Орбиты относятся как {orbit_1 // 3}:{orbit_2 // 3}.')
elif orbit_1 % 2 == 0 and orbit_2 % 2 == 0:
    print(f'Орбиты относятся как {orbit_1 // 2}:{orbit_2 // 2}.')
else:
    print('Орбиты не синхронизированы.')
```
