```python
"""
Пусть нам нужно собрать монетки, пока не наберём требуемую сумму в 50 рублей. Решим такую задачу с помощью цикла.

Заведём переменные: s — для подсчёта суммы, начальное значение 0; coin — для введённого номинала монетки, запишем в неё первое число, которое введёт пользователь.

Теперь нужно сформулировать условие входа в цикл: если мы ещё не набрали необходимую сумму, то в цикл входим.

В цикле будем выполнять следующие действия:

Добавим номинал монетки к сумме.
Введём новую монетку.
Наша программа не остановится, пока не наберётся требуемая сумма, поэтому нужно гарантировать, что вводимых данных для этого достаточно. А вот остальные монетки, если такие будут, не будут учитываться, сколько бы их ни было введено.
"""

s = 0

while s < 50:
    coin = int(input())
    s += coin
print(s)
```

```python
s = 0
coin = int(input())
while s < 50:
    s = s + coin
    coin = int(input())
print(s)
```

```python
"""
Найдите количество неотрицательных чисел среди введённых пользователем.

Формат ввода
Числа считываются, пока не будет введено число по модулю большее 1000.

Оно в подсчётах не участвует, служит только условием окончания ввода.

Формат вывода
Выведите количество неотрицательных чисел среди прочитанных до наступления условия.
"""
a = int(input())
count = 0

while abs(a) <= 1000:
    a = int(input())
    if a >= 0:
        count = count + 1
        print(count)
```

```python
a = int(input())
count = 0

while abs(a) <= 1000:
    if a >= 0:
        count += 1
    a = int(input())

print(count)
```

```python
a = float(input())
count = 0
summ = 0

while a != -1:
    summ = a + summ
    count += 1
    avg = summ / count
    a = float(input())
print(avg)
```

```python
n = int(input())

while n <= 9999:
    d1 = n // 1000
    d2 = n // 100 % 10
    d3 = n // 10 % 10
    d4 = n % 10

    print(d1, d2, d3, d4, sep='+')
    n = int(input())
```
