```python
"""
Напишите программу, которая читает три строки, а затем выводит
их в таком порядке: первую, затем удвоенную вторую, затем третью.
Для удвоения обязательно используйте умножение строк.
"""
s1 = str(input())
s2 = str(input())
s3 = str(input())

print(s1 + ' ' + 2 * s2 + ' ' + s3)
```

```python
"""
Кодом числа назовём его преобразование в такой вид:
сначала запишем остаток от деления числа на 2, к нему справа припишем остаток от деления числа на 3, затем на 4 и, наконец, на 5.

Напишите программу, которая по введённому числу получит и выведет его код. Помните, что числа не могут начинаться с нуля.
Для склеивания чисел может потребоваться превращение их в строки.
"""

a = int(input())

b_str = str(a % 2)
c_str = str(a % 3)
d_str = str(a % 4)
e_str = str(a % 5)
t = b_str + c_str + d_str + e_str
print(int(t))
```
