## _For_ Loop

```python
colors = ["red", "orange", "yellow", "green", "blue", "indigo", "violet"]
```

```python
for color in colors:
    print(color)
```

```python
for number in range(1, 5):
    print(number)
```

```python
for number in range(5):
    print(number)
```

```python
fruits = ["apple", "banana", "orange"]
fruits
```

```python
for index, fruit in enumerate(fruits):
    print(f"At position {index}, I found a {fruit}")
```

```python
# For loop example

dates = [1982,1980,1973]
N = len(dates)

for i in range(N):
    print(dates[i])
```

```python
# Exmaple of for loop, loop through list

for year in dates:
    print(year)
```

```python
# Use for loop to change the elements in list

squares = ['red', 'yellow', 'green', 'purple', 'blue']

for i in range(0, 5):
    print("Before square ", i, 'is',  squares[i])
    squares[i] = 'white'
    print("After square ", i, 'is',  squares[i])
```

```python
# Loop through the list and iterate on both index and element value
# enumerate() превращает список в последовательность пар:
# (индекс, элемент)

squares=['red', 'yellow', 'green', 'purple', 'blue']

for i, square in enumerate(squares): # распаковка tuple
    # Каждая пара разбивается на две переменные:
    # i → индекс
    # square → значение элемента
    print(i, square)

# в случае enumerate python видит:
# (0, 'red')
# (1, 'yellow')
# (2, 'green')
# (3, 'purple')
# (4, 'blue')
```

```python
# без enumerate

for i in range(len(squares)):
    print(i, squares[i])
```

```python
# можно менять стартовый индекс

for i, square in enumerate(squares, start=1):
    print(i, square)
```

## _While_ Loop

```python
"""while condition:
    # Code to be executed while the condition is true
    # Indentation is crucial to indicate the scope of the loop
"""
count = 1

while count <= 4:
    print(count)
    count += 1
```

```python
# While Loop Example

dates = [1982, 1980, 1973, 2000]

i = 0
year = dates[0]

while(year != 1973):
    print(year)
    i = i + 1
    year = dates[i]


print("It took ", i ,"repetitions to get out of loop.")
```

```python
# 1) create the source list
squares = ['orange', 'orange', 'orange', 'blue', 'orange']

# 2) create an empty list where we will copy the values
new_squares = []

# 3) initialize the index for the while loop
i = 0

# 4) loop through the list while the index is within the list length
while i < len(squares):

    # 5) check if the current element is 'orange'
    if squares[i] == 'orange':

        # 6) copy the value to the new list
        new_squares.append(squares[i])

        # 7) move to the next element
        i += 1

    else:
        # 8) stop the loop if the value is not 'orange'
        break

# 9) print the result
print(new_squares)
```

```python
squares = ['orange', 'orange', 'purple', 'blue ', 'orange']
new_squares = []
i = 0
while(i < len(squares) and squares[i] == 'orange'):
    new_squares.append(squares[i])
    i = i + 1
print (new_squares)
```

### When to Use Each

#### _For Loops:_ Use for loops when you know the number of iterations in advance and want to process each element in a sequence. They are best suited for iterating over collections and sequences where the length is known.

#### _While Loops:_ Use while loops when you need to perform a task repeatedly as long as a certain condition holds true. While loops are particularly useful for situations where the number of iterations is uncertain or where you're waiting for a specific condition to be met.

## Controlling Loop Execution with <code>break</code> and <code>continue</code>

#### Break


```python
for num in range(1, 10):
    if num == 7:
        print("Breaking the loop at:", num)
        break
    print(num)
```

#### Continue

```python
for num in range(1, 6):
    if num == 3:
        continue
    print(num)
```

```python
count = 0
while count < 10:
    count += 1
    if count == 3:
        continue  # skip printing 3
    if count == 8:
        break     # stop the loop when count is 8
    print(count)
```
