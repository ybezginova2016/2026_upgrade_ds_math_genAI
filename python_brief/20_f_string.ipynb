```python
pi = 22 / 7
print(F'Fraction value is approximately equal to {pi:.5f} or {pi:.3f} or {pi:.0f}.')
```

```python
"""
f'...' означает: внутри строки можно подставлять значения переменных
{year:04d} — вывести year как целое число из 4 цифр
{month:02d} — вывести month как целое число из 2 цифр
{day:02d} — вывести day как целое число из 2 цифр
"""

year = int(input())
month = int(input())
day = int(input())
print(f'Date {year:04d}.{month:02d}.{day:02d}')
```

```python
"""
вывести строку name
выделить под нее минимум 8 символов
"""

name = input()
age = int(input())
print(f'{name:8} -> {age:5d}')

```

```python
"""
Соберите строку, в которой введённое пользователем название конторы выводится в правильных кавычках.

Сначала вводится число 1 или 2, которое означает тип кавычек (одинарные или двойные), затем название конторы, которое нужно вставить в строку:

Черноморское отделение Арбатской конторы <название конторы в кавычках>.
"""
t = int(input())
name = str(input())

if t == 1:
    print(f"Черноморское отделение Арбатской конторы '{name}'.")
else:
    print(f'Черноморское отделение Арбатской конторы "{name}".')
```

```python
"""
Проверьте, есть ли введённая первой подстрока в строке, введённой следом? Если есть, то так и выведите:
Есть! А если нет, то выведите:
Спряталась.
"""

s1 = str(input())
s2 = str(input())

if s1 in s2:
    print('Есть!')
else:
    print('Спряталась.')
```
