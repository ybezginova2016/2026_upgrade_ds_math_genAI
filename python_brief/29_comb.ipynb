```python
"""
##### ПОРЯДОК НЕ ВАЖЕН #####

Из класса в n учеников нужно выбрать k человек для украшения актового зала. Определите, сколькими способами это можно сделать? Группы, различающиеся порядком учеников, считаются одинаковыми.
"""
from math import perm

n = int(input())
k = int(input())

print(perm(n, k))
```

```python
"""
Биты, байты, килобайты

Для хранения информации в памяти отводится целое число байтов или килобайтов. А сама информация измеряется битами. Например, если для хранения информации об одном символе нужно i бит, а таких символов в сообщении n, то сами понимаете, сколько потребуется битов для всего сообщения.

Напишите программу, которая определит наименьшее целое число байтов или килобайтов для хранения сообщения из n символов, каждый из которых записывается i битами.

Если килобайты или байты не нужны, их выводить не нужно.

Формат ввода
Вводятся два числа: n и i.

Формат вывода
Выведите ответ на задачу.
"""
from math import ceil

n = float(input())
i = float(input())

bits = n * i
bytes_count = ceil(bits / 8)

kilobytes = bytes_count // 1024
bytes_left = bytes_count % 1024

if kilobytes > 0 and bytes_left > 0:
    print(kilobytes, bytes_left)
elif kilobytes > 0:
    print(kilobytes)
else:
    print(bytes_count)
```

```python
from math import sin, radians

alpha = float(input())
a = float(input())
b = float(input())
c = float(input())

sin_alpha = sin(radians(alpha))

sin_beta = b * sin_alpha / a
sin_gamma = c * sin_alpha / a

print(sin_beta, sin_gamma)
```

```python
a = float(input())
b = float(input())
c = float(input())

d = b ** 2 - 4 * a * c

if d < 0:
    print('None')
elif d == 0:
    x = -b / (2 * a)
    print(round(x, 5))
else:
    x1 = (-b - d ** 0.5) / (2 * a)
    x2 = (-b + d ** 0.5) / (2 * a)
    print(round(x1, 5), round(x2, 5))
```
