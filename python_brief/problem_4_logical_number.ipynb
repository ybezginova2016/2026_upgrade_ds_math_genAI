```python
"""
Назовём число отличным, если оно делится на 3 и 5. Если только на 5, то назовём его полукруглым. Если только на 3, то тройным, а если не кратно ни 3, ни 5, то пусть будет просто числом.

Напишите программу, которая выведет название числа (отличное, полукруглое, тройное или просто число) в зависимости от его кратности.
"""

exc_num = int(input())

if exc_num % 3 and exc_num % 5:
    print('отличное')
elif exc_num % 5:
    print('полукруглым')
elif exc_num % 3:
    print('тройным')
else:
    print('просто число')
```
