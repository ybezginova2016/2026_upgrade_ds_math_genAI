```python
"""
Всё, что сверху

Напишите программу для проверки попадания числа в некоторый диапазон.

Формат ввода
Вводятся два числа — границы диапазона в произвольном порядке. Затем вводится число для проверки.

Формат вывода
Определите, где находится введённое число — ниже нижнего, выше верхнего, внутри диапазона или на какой-либо границе. И выведите соответственно:

ниже, выше, внутри, граница

Пример 1
Ввод	Вывод
5
7
12
выше
"""

a = int(input())
b = int(input())
n = int(input())

left = min(a, b)
right = max(a, b)

if n < left:
    print('ниже')
elif n > right:
    print('выше')
elif n == left or n == right:
    print('граница')
else:
    print('внутри')
```

```python
"""
Перестановки цифр

Из введённого трёхзначного числа получите наибольшее трёхзначное число, составленное из тех же цифр.

Пример 1
Ввод	Вывод
103
310

"""

n = int(input())

a = n // 100
b = n // 10 % 10
c = n % 10

if a >= b and a >= c:
    first = a
    if b >= c:
        second = b
        third = c
    else:
        second = c
        third = b
elif b >= a and b >= c:
    first = b
    if a >= c:
        second = a
        third = c
    else:
        second = c
        third = a
else:
    first = c
    if a >= b:
        second = a
        third = b
    else:
        second = b
        third = a

print(first * 100 + second * 10 + third)
```
