### GET request

```python
url_get = 'https://httpbin.org/get'
payload = {'name': 'Joseph',
           'ID': '123'}
```

```python
r = re.get(url_get, params=payload)
r.url
```

### Post Requests

```python
url_post = 'https://httpbin.org/post'
```

```python
payload = {
    'name': 'Joseph',
    'ID': '123'
}
```

```python
r_post = re.post(url_post, data = payload)
r_post
```

```python
# compare POST and GET

print('POST request URL:', r_post.url)
print('GET request "URL:', r.url)
```

```python
# compare POST and GET body

print('POST request body', r_post.request.body)
print('GET request body', r.request.body)
```

```python
r_post.json()
```
